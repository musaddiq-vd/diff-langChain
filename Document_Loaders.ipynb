# PDF Loader (.pdf)

```python
from langchain_community.document_loaders import PyPDFLoader

# Loader: Loads PDF file
loader = PyPDFLoader("sample.pdf")

# Load: Reads PDF page by page
documents = loader.load()

# Print loaded content
print(documents)
```

# Text Loader (.txt)

```python
from langchain_community.document_loaders import TextLoader

# Loader: Loads text file
loader = TextLoader("sample.txt")

# Load: Reads file and returns Document objects
documents = loader.load()

# Print loaded content
print(documents)
```

# CSV Loader (.csv)

```python
from langchain_community.document_loaders import CSVLoader

# Loader: Loads CSV file
loader = CSVLoader(file_path="data.csv")

# Load: Reads CSV rows as documents
documents = loader.load()

# Print loaded content
print(documents)
```

# Web Page Loader

```python
from langchain_community.document_loaders import WebBaseLoader

# Loader: Loads web page content
loader = WebBaseLoader("https://example.com")

# Load: Reads webpage text
documents = loader.load()

# Print loaded content
print(documents)
```

# Directory Loader

```python
from langchain_community.document_loaders import DirectoryLoader
from langchain_community.document_loaders import PyPDFLoader

# DirectoryLoader: Loads multiple files from a folder
loader = DirectoryLoader(
    "docs",                 # Folder containing PDFs
    glob="*.pdf",           # Load only PDF files
    loader_cls=PyPDFLoader  # Use PyPDFLoader for each PDF
)

# Load: Reads all PDFs and returns Document objects
docs = loader.load()

# Print loaded documents
print(docs)
```
